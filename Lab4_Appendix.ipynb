```python
LAST_NAME = "Lim"
STUDENT_ID = "TUPM-26-3484"
FAVORITE_ARTIST = "BRENT FAIYAZ"

seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)

sys_config = {
    "operator": LAST_NAME,
    "auth_id": STUDENT_ID,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED"
}

base_val = sys_config["base_seed"]

char_values = [ord(char) + base_val for char in FAVORITE_ARTIST if char.strip()]

print("=== SYSTEM CONFIGURATION ===")
for key, value in sys_config.items():
    print(f"{key.upper()}: {value}")
print(f"\nBase Values Generated from Artist: {char_values}")
```

```python
# [EXERCISE 1] Engineering Test Data Organizer
from collections import defaultdict
import statistics

sources = [f"{sys_config['operator']}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

raw_test_data = [
    (sources[0], categories[0], char_values[0]),
    (sources[1], categories[1], char_values[1]),
    (sources[0], categories[0], char_values[0]),
    (sources[2], categories[2], char_values[2]),
    (sources[3], categories[3], char_values[3]),
    (sources[4], categories[4], char_values[4]),
    (sources[0], categories[1], char_values[5]),
    (sources[1], categories[2], char_values[6])
]

organized_data = defaultdict(list)
seen_records = set()
repeated_records = set()

for record in raw_test_data:
    if record in seen_records:
        repeated_records.add(record)
    else:
        seen_records.add(record)
        organized_data[(record[0], record[1])].append(record[2])

summary = {}
for (src, cat), vals in organized_data.items():
    summary[(src, cat)] = {
        "Max": max(vals),
        "Min": min(vals),
        "Average": statistics.mean(vals) if vals else 0
    }

print("=== ENGINEERING TEST DATA ORGANIZER ===")
print(f"\nRepeated Records Detected: {repeated_records}")
print("\nConsolidated Final Summary:")
for key in sorted(summary.keys()):
    print(f"Source: {key[0]} | Category: {key[1]} | Stats: {summary[key]}")
```

```python
# [Exercise 2] Student Performance Data Analyzer
students = [f"{sys_config['operator']}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

raw_performance_data = [
    (students[0], subjects[0], char_values[0]),
    (students[1], subjects[1], char_values[1]),
    (students[0], subjects[0], char_values[0]),
    (students[2], subjects[2], None),
    (students[3], subjects[3], char_values[3]),
    (students[4], subjects[4], char_values[4])
]

def get_category(score):
    if score is None: return "INCOMPLETE"
    if score >= 85: return "EXCELLENT"
    if score >= 75: return "PASS"
    return "FAIL"

unique_perf_data = set()
repeated_perf = set()
incomplete_perf = []
final_report = []

for record in raw_performance_data:
    if record in unique_perf_data:
        repeated_perf.add(record)
    else:
        unique_perf_data.add(record)
        if record[2] is None:
            incomplete_perf.append(record)
        
        category = get_category(record[2])
        final_report.append({"Student": record[0], "Subject": record[1], "Score": record[2], "Category": category})

print("=== STUDENT PERFORMANCE REPORT ===")
print(f"Repeated Records: {repeated_perf}")
print(f"Incomplete Records: {incomplete_perf}")
print("\nFinal Analyzed Results (Ordered by Student):")
for entry in sorted(final_report, key=lambda x: x["Student"]):
    print(entry)
```

```python
# [EXERCISE 3] Transaction History Analyzer
customers = [f"{sys_config['operator']}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 50, "PEN": 20, "USB": 300, "HEADPHONES": 500, "CALCULATOR": 800}

raw_transactions = [
    (customers[0], products[0], char_values[0]),
    (customers[1], products[1], char_values[1]),
    (customers[0], products[0], char_values[0]),
    (customers[2], products[2], char_values[2]),
    (customers[3], products[3], char_values[3]),
    (customers[4], products[4], char_values[4])
]

transaction_history = set()
repeated_transactions = set()
customer_summary = defaultdict(int)
product_summary = defaultdict(int)
overall_value = 0

for t in raw_transactions:
    if t in transaction_history:
        repeated_transactions.add(t)
    else:
        transaction_history.add(t)
        
        total_price = prices[t[1]] * t[2]
        overall_value += total_price
        
        customer_summary[t[0]] += total_price
        product_summary[t[1]] += t[2]

print("=== TRANSACTION HISTORY REPORT ===")
print(f"Repeated Purchases Detected: {repeated_transactions}")
print(f"\nOverall Dataset Transaction Value: P{overall_value}")

print("\nCustomer Spending Summary:")
for cust, total in sorted(customer_summary.items()):
    print(f"{cust}: P{total}")

print("\nProduct Units Sold Summary:")
for prod, qty in sorted(product_summary.items()):
    print(f"{prod}: {qty} units")
```
